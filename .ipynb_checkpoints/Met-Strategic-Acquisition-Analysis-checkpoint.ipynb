```python
# 1. Define the web address (URL) for the Met's department list endpoint
# 2. Send a request to that URL and store whatever comes back
# 3. Check that the request succeeded (status code 200)
# 4. Convert the response into Python dictionary to read it
# 5. Print it to see the department names and their ID numbers
```

```python
import requests

url = "https://collectionapi.metmuseum.org/public/collection/v1/departments"
response = requests.get(url)
response.status_code
```

```python
departments = response.json()
departments
```

```python
objects_url = "https://collectionapi.metmuseum.org/public/collection/v1/objects"
params = {"departmentIds": 21}
response = requests.get(objects_url, params=params)
response.status_code
```

```python
modern_art_objects = response.json()
modern_art_objects.keys()

#checking for keys, so I don't get dumped with thousands of object IDs
```

```python
modern_art_objects["total"]

#checking for total objects in the modern art collection
```

```python
import random

object_ids = modern_art_objects["objectIDs"]
sample_ids = random.sample(object_ids, 1200)

# pulling a moderate random sample, 15,135  separate requests would hit rate limits on API
# 1200 randomly selected object IDs will give statistically meaningful nationality/gender distributions without the runtime cost
```

```python
len(sample_ids)
```

```python
import time

object_details = []

for i, object_id in enumerate(sample_ids): # gives both the item and its position number as it loops
    obj_url = f"https://collectionapi.metmuseum.org/public/collection/v1/objects/{object_id}"

    success = False
    attempts = 0

    
    obj_response = requests.get(obj_url)

    try:
        obj_data = obj_response.json()
        object_details.append(obj_data)
    except ValueError:  
        print(f"Skipped object {object_id} - bad response") # wrap each request in a try/except so one bad object doesn't kill the whole loop

    if (i + 1) % 100 == 0:
        print(f"Pulled {i + 1} of {len(sample_ids)} objects")  # to check the count every 100th object to see if the code is running

    time.sleep(0.1) # small pause between each request to reduce failures if rate-limiting is the cause

len(object_details)
```

```python
len(object_details)
```

```python
import pandas as pd

met_df = pd.DataFrame(object_details) # convert the list of dicts into dataframe

met_df.shape
met_df.columns
met_df.head()
```

```python
list(met_df.columns)
```

```python
columns_needed = ["objectID", "artistDisplayName", "artistNationality",
                  "artistGender", "accessionYear", "objectName",
                  "department", "culture"]

met_clean = met_df[columns_needed]
met_clean.shape
```
